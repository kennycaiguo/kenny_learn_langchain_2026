## pip install langchain-community安装langchain的社区包
## 获取智普GLM-5.3-Flash大模型的api key并且保存到.env中，同时需要保存base url
## 官网：https://bigmodel.cn/

```python
from langchain_community.chat_models import ChatZhipuAI
from dotenv import load_dotenv
import os

load_dotenv(override=True)
zhipuai_api_key = os.getenv("ZHIPUAI_API_KEY")
zhipuai_base_url = os.getenv("ZHIPUAI_BASE_URL")

llm=ChatZhipuAI(
    api_key=zhipuai_api_key,
    api_base=zhipuai_base_url,
    model="glm-5.3-flash"
)

# 调用模型
resp= llm.invoke("阿里巴巴是一家什么样的公司")
print(resp.text)
```

# ChatOpenAI兼容模式运行智普大模型

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os

load_dotenv(override=True)
ZHIPUAI_API_KEY = os.getenv("ZHIPUAI_API_KEY")
ZHIPUAI_BASE_URL = os.getenv("ZHIPUAI_BASE_URL")

llm = ChatOpenAI(
    model="glm-5.3-flash",
    api_key=ZHIPUAI_API_KEY,
    base_url=ZHIPUAI_BASE_URL
)

resp = llm.invoke("奇门遁甲是什么")
print(resp.text)
```

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os

load_dotenv(override=True)
CHATGPT_API_KEY = os.getenv("CHATGPT_API_KEY")

llm = ChatOpenAI(
    model="gpt-5.4-mini",
    api_key=CHATGPT_API_KEY
)

resp = llm.invoke("奇门遁甲是什么")
print(resp.text)
```
