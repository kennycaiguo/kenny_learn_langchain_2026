## 首先需要获取一个groq的apikey，链接：https://console.groq.com/docs/text-chat
注意不要和别人分享这个key
需要在系统的环境变量里面创建一个GROQ_API_KEY，然后把密钥的值保存在里面
## 然后需要安装groq库：pip install groq
# 注意：llama-3.3-70b-versatile已经不能够使用了。可以使用：openai/gpt-oss-120b

```python
from groq import Groq

client = Groq()
chat = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[{
        "role":"user","content":"please write a java helloworld program"
    }]
)

print(chat.choices[0].message.content)
```
