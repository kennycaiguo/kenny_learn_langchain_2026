## 首先需要获取一个groq的apikey，链接：https://console.groq.com/docs/text-chat
注意不要和别人分享这个key
需要在系统的环境变量里面创建一个GROQ_API_KEY，然后把密钥的值保存在里面
## 然后需要安装groq库：pip install groq
# 注意：llama-3.3-70b-versatile已经不能够使用了。可以使用：openai/gpt-oss-120b

```python
from groq import Groq

client = Groq()
chat = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[{
        "role":"user","content":"please write a java helloworld program"
    }]
)

print(chat.choices[0].message.content)
```

```python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv(override=True)
GROQ_API_KEY=os.getenv("GROQ_API_KEY")

# 初始化客户端并发送对话请求
client = OpenAI(
    api_key=GROQ_API_KEY,
    base_url="https://api.groq.com/openai/v1"
)

response = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[{"role": "user", "content": "你好"}]
)
print(response.choices[0].message.content)
```

```python
import os
from langchain_openai import ChatOpenAI


# 初始化 ChatOpenAI 调用 Groq 模型
llm = ChatOpenAI(
    model="openai/gpt-oss-120b",  # 或者是 Groq 支持的其他模型名字
    temperature=0.7,
    api_key=os.environ.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1",  # 指向 Groq 的 OpenAI 兼容接口
)

# 调用模型
response = llm.invoke("你好法语怎么说")
print(response.content)
```

```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage


llm = init_chat_model(
    model="openai/gpt-oss-120b",  # 指定具体的 Groq 模型名称
    model_provider="groq",    # 明确指定提供商为 groq
    temperature=0.7
)

# 调用模型
messages = [
    # HumanMessage(content="你好的德文"),
    HumanMessage(content="圣荷西今天的天气"),
]

response = llm.invoke(messages)

# 打印输出结果
print(response.content)
```
