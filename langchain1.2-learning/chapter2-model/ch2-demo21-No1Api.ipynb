## No1API

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage,SystemMessage
import os

load_dotenv(override=True)
No1API_KEY = os.getenv("No1API_KEY")
No1_BASE_URL = os.getenv("No1_BASE_URL")

llm = ChatOpenAI(
    model="gpt-3.5-turbo",
    api_key=No1API_KEY,
    base_url=No1_BASE_URL
)

messages = "性交的荷兰语"

resp = llm.invoke(messages)
resp.pretty_print()
```
