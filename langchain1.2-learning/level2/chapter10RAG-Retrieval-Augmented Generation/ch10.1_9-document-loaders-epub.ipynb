# 加载epub
## 1.需要安装pypandoc： pip install pypandoc
##  2.操作系统需要安装pandoc，在这里下载： https://github.com/jgm/pandoc/releases
### pypandoc 1.17 内置或默认下载对应的 Pandoc 版本是 3.9，所以这里我们需要下载3.9版本

```python
from langchain_community.document_loaders import UnstructuredEPubLoader
from pprint import pprint
loader = UnstructuredEPubLoader(
    file_path="../asset/load/13-半小时漫画经济学4_ 理财篇.epub",
    # mode: "single","elements","paged"
    mode="single"
)

docs = loader.load()
print(len(docs))
pprint(docs)
```
