# 加载excel文件
## 需要安装openpyxl 和xlrd  ：pip install openpyxl xlrd
## 还需要安装msoffcrypto： pip install msoffcrypto-tool
## 情况1，mode="single",strategy="fast"

```python
from langchain_community.document_loaders import UnstructuredExcelLoader
from pprint import pprint

loader = UnstructuredExcelLoader(
    file_path="../../asset/load/10-load.xlsx",
    mode="single",
    strategy="fast"
)

docs = loader.load()
print(len(docs))
pprint(docs)
```

## 情况2. mode="elements",strategy="fast"

```python
from langchain_community.document_loaders import UnstructuredExcelLoader
from pprint import pprint

loader = UnstructuredExcelLoader(
    file_path="../../asset/load/10-load.xlsx",
    mode="elements",
    strategy="fast"
)

docs = loader.load()
print(len(docs))
pprint(docs)
```
