# 文档加载器 Document Loaders
## 3.加载json文件
### 需要安装jq： pip install jq
### jq的用法
常见 jq schema 参考：

JSON -> [{"text": ...}, {"text": ...}, {"text": ...}] jq_schema -> ".[].text"

JSON -> {"key": [{"text": ...}, {"text": ...}, {"text": ...}]} jq_schema -> ".key[].text"

JSON -> ["...", "...", "..."] jq_schema -> ".[]"
### 案例1情况1,
#### jq_schema=".", #直接提取完整的JSON对象（包括所有字段）
#### text_content=False #保持原始 JSON 结构，将提取的数据转换为JSON字符串存入page_content字段中

```python
from langchain_community.document_loaders import JSONLoader
from rich import print as rprint
loader = JSONLoader(
    file_path="../../asset/load/03-load.json",
    jq_schema=".",
    text_content=False
)

docs = loader.load()
rprint(docs)
```

### 案例1情况2. .messages[].content:遍历.messages[]中所有元素 从每一个元素中提取.content字段


```python
from langchain_community.document_loaders import JSONLoader
from rich import print as rprint
loader = JSONLoader(
    file_path="../../asset/load/03-load.json",
    jq_schema=".messages[].content",
)

docs = loader.load()
rprint(docs)
```

## 案例2情况3：提取data.items中指定字段的数据

```python
from langchain_community.document_loaders import JSONLoader
from rich import print as rprint

file_path="../../asset/load/03-response.json"

loader = JSONLoader(
    file_path=file_path,  # 文件路径
    jq_schema="""
        .data.items[] | {
            author,
            created_at,
            content: (.title + "\n" + .content)
        }
    """,
    text_content=False,  # 提取内容是否为字符串格式
)

docs = loader.load()
rprint(docs)
```
