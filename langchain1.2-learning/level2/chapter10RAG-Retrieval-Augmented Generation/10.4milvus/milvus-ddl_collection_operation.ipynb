# Milvus数据库操作DDL

### 举例1.创建数据库并且切换数据库
#### milvus数据库的默认端口是19530

```python
from pymilvus import MilvusClient
from rich import print as rprint

# 创建客户端对象
client = MilvusClient("http://localhost:19530")

# 新建一个rag_tutorial数据库
client.create_database(db_name="rag_tutorial")

# 查看一下
dbs = client.list_databases()
# rprint(dbs) # ['default', 'rag_tutorial']
# 切换到数据库reg_totorial
client.use_database("rag_tutorial")

```

### 举例2查看数据库下面的collection

```python
colls = client.list_collections()
rprint(colls)
```

### 举例3.创建collection


```python
col_name = "docs"
client.create_collection(
    collection_name=col_name,
    dimension=1024, # 注意：这里的数据维度必须和我们需要使用的嵌入模型支持的维度一样
    metric_type="COSINE",
)
```

```python
rprint(client.list_collections())
```

### 举例4 删除collection

```python
client.drop_collection(collection_name=col_name)
rprint(client.list_collections())
```
