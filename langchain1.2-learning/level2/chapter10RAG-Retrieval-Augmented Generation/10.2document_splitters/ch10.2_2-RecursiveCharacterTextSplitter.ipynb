# RecursiveCharacterTextSplitter：最常用¶
## 常用参数
### chunk_size,
### chunk_overlap,
### add_start_index,注意，只有开始的index,没有结束的index
## 举例1：使用split_text()方法演示

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1,定义RecursiveCharacterTextSplitter分割器对象
splitter = RecursiveCharacterTextSplitter(
    chunk_size=10,
    chunk_overlap=0,
    add_start_index=True,
)
# 2.定义被分割内容
text="LangChain框架特性\n\n多模型集成(GPT/Claude)\n记忆管理功能\n链式调用设计。文档分析场景示例：需要处理PDF/Word等格式。"

paragraphs = splitter.split_text(text)
for i,chunk in enumerate(paragraphs):
    print(f"块{i + 1},长度：{len(chunk)}")
    print(chunk)
    print('-' * 50)
```

## 2.使用使用create_documents()方法演示，传入字符串列表，返回Document对象列表

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1,定义RecursiveCharacterTextSplitter分割器对象
splitter = RecursiveCharacterTextSplitter(
    chunk_size=10,
    chunk_overlap=0,
    add_start_index=True,
)

txt_list=["LangChain框架特性\n\n多模型集成(GPT/Claude)\n记忆管理功能\n链式调用设计。文档分析场景示例：需要处理PDF/Word等格式。"]

docs = splitter.create_documents(texts=txt_list)
for doc in docs:
    print(doc)
    print('-' * 50)
```

## 举例3：使用create_documents()方法演示，将本地文件内容加载成字符串，进行拆分

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

## 打开文本读取内容，保存到一个变量
with open("../asset/load/09-ai.txt", "r", encoding="utf-8") as f:
    content = f.read() # 这个是一个字符串
splitter = RecursiveCharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=20, ## 注意。这个值一定要比chunk_size小
    length_function=len
)

texts = splitter.create_documents([content])
# 5.打印分割文本
for text in texts:
    print(f"🔥{text.page_content}")
```

## 举例3_2：使用create_documents()方法演示，将本地文件内容加载成字符串列表，进行拆分

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

## 打开文本读取内容，保存到一个变量
with open("../asset/load/09-ai.txt", "r", encoding="utf-8") as f:
    lines = f.readlines() # 这个就是一个字符串列表
splitter = RecursiveCharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=20, ## 注意。这个值一定要比chunk_size小
    length_function=len
)

texts = splitter.create_documents(lines)
# 5.打印分割文本
for text in texts:
    print(f"🔥{text.page_content}")
```

## 举例4：使用split_documents()方法演示，利用PDFLoader加载文档，对文档的内容用递归切割器切割

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

loader = PyPDFLoader(file_path="../asset/load/04-load.pdf")
docs = loader.load()

splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=0,
    length_function=len,
    add_start_index=True,
)

texts = splitter.split_documents(docs)
for text in texts:
    print(text)
    print("="*50)
```

## 举例5.使用split_documents()方法演示，利用UnstructuredEPubLoader加载epub文档，对文档的内容用递归切割器切割

```python
from langchain_community.document_loaders import UnstructuredEPubLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

loader = UnstructuredEPubLoader(file_path="../asset/load/13-半小时漫画经济学4_ 理财篇.epub")
docs = loader.load()
splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=0,
    length_function=len,
    add_start_index=True,
)
texts = splitter.split_documents(docs)
for text in texts:
    text=text.page_content.replace("\n","") ## 消除里面的换行符
    print(text)
    print("="*50)
```
