# CodeTextSplitter
## CodeTextSplitter（或称代码文本分割器）是用于将源代码按语言特定语法（如类、函数、方法定义等）进行结构化拆分的工具，最常见于 LangChain 文本分割集成指南。

### 核心特点与工作原理

#### • 结构感知（Structure-Aware）：与按固定字符或段落强行截断不同，代码分割器会根据编程语言的语法边界（例如 Python 的 class、def 或大括号等）来切分代码，避免破坏函数或类内部的逻辑完整性。
#### • 推荐用法：在现代 LangChain 框架中，官方更推荐使用通用且支持多语言的 RecursiveCharacterTextSplitter.from_language() 方法，通过传入语言枚举（如 Language.PYTHON、Language.JS 等）来获取针对该语言优化过的分隔符

## 1.查看支持的编程语言

```python
from langchain_text_splitters import Language

# 支持分割语言类型
# Full list of supported languages
langs = [e.value for e in Language]
print(langs)
```

## 使用通用且支持多语言的 RecursiveCharacterTextSplitter.from_language() 方法

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from pprint import pprint

# 2.定义要分割的python代码片段
PYTHON_CODE = """
def hello_world():
    print("Hello, World!")

def hello_world1():
    print("Hello, World1!")
"""

splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON,
    chunk_size=50,
    chunk_overlap=0,
)

codes = splitter.create_documents(texts=[PYTHON_CODE])
pprint(codes)
```

### demo2

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language

# 1. 定义一段待分割的 Python 源代码
PYTHON_CODE = """
def add(a, b):
    return a + b

class Calculator:
    def __init__(self):
        self.result = 0

    def multiply(self, a, b):
        return a * b
"""

# 2. 实例化针对 Python 语言的代码分割器
# chunk_size: 每个代码块的最大字符数
# chunk_overlap: 两个相邻代码块之间的重叠字符数
python_splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON,
    chunk_size=120,
    chunk_overlap=0
)

# 3. 执行代码分割
docs = python_splitter.create_documents([PYTHON_CODE])

# 4. 查看分割结果
for i, doc in enumerate(docs):
    print(f"--- Chunk {i+1} ---")
    print(doc.page_content)
```
