# Token文本分割器： TokenTextSplitter
## 用token来分割文本


### 举例1
### 常用参数
#### chunk_size=33,  # 最大 token 数为 33
#### chunk_overlap=0, # 重叠 token 数为 0
#### (model_name="gpt-4", # 选择 GPT-4 模型的编码器,这个参数尽量少用)
#### encoding_name="cl100k_base",  # 使用 OpenAI 的编码器,将文本转换为 token 序列

```python
from langchain_text_splitters import TokenTextSplitter

splitter = TokenTextSplitter(
    chunk_size=33,
    chunk_overlap=0,
    encoding_name='cl100k_base'
)

text = "人工智能是一个强大的开发框架。它支持多种语言模型和工具链。人工智能是指通过计算机程序模拟人类智能的一门科学。自20世纪50年代诞生以来，人工智能经历了多次起伏。"

texts = splitter.split_text(text)

print(f"原始文本被分成了{len(texts)}块")
for i,chunk in enumerate(texts):
    print(f"块{i+1},内容：{chunk},块长度：{len(chunk)}")
    print("*"*50)
```

### 举例2.使用CharacterTextSplitter.from_tiktoken_encoder来分割


```python
from langchain_text_splitters import CharacterTextSplitter

splitter = CharacterTextSplitter.from_tiktoken_encoder(
    encoding_name='cl100k_base',
    chunk_size=18,
    chunk_overlap=0,
    separator='。',
    keep_separator=False
)

text = "人工智能是一个强大的开发框架。它支持多种语言模型和工具链。今天天气很好，想出去踏青。但是又比较懒不想出去，怎么办"
texts = splitter.split_text(text)
print(f"原始文本被分成了{len(texts)}块")
for i,chunk in enumerate(texts):
    print(f"块{i+1},内容：{chunk},块长度：{len(chunk)}")
    print("*"*50)
```
