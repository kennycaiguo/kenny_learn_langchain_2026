# 长期记忆1.基本API，put()/get()
## 1.InMemoryStore

```python
from langgraph.store.memory import InMemoryStore

store = InMemoryStore()
namespace=("users",)
user_id = "user_1"
user_name="Jack"
store.put(namespace, user_id, {"name":user_name})
item = store.get(namespace, user_id)
print(item)
```

### update data

```python
user_name = "Mary"
store.put(namespace, user_id, {"name":user_name})
item = store.get(namespace, user_id)
print(item)
```

## 2、put()/get()基于PostgresStore

```python
from langgraph.store.postgres import PostgresStore
from dotenv import load_dotenv
import os

load_dotenv() ## 加载.env文件，因为我们的数据库用户和密码都在里面，然后分别加载他们
db_user = os.getenv("PG_USER")
db_pwd = os.getenv("PG_PWD")
# 拼接数据库url
db_url = f"postgresql://{db_user}:{db_pwd}@localhost:5432/langchain_db?sslmode=disable"

with PostgresStore.from_conn_string(db_url) as store:
     # 数据库的store需要setup
     store.setup()
     namespace=("users",)
     user_id = "user_1"
     user_name="Jack"
     store.put(namespace, user_id, {"name":user_name})
     item = store.get(namespace, user_id)
     print(item)
```

### update data

```python
with PostgresStore.from_conn_string(db_url) as store:
     store.setup() # 每一次调用都需要打开连接 和setup
     user_name="Brenda"
     store.put(namespace, user_id, {"name":user_name})
     item = store.get(namespace, user_id)
     print(item)
```
