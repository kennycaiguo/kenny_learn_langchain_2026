```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

# 实例化提示模板
tmp = ChatPromptTemplate.from_messages([
  ("system","你是一个非常友好的Ai助手,你的名字叫{name}"),
  ("user","你好，最近怎么样？"),
  ("ai","我很好，谢谢"),
  ("human","{user_input}")
])

# 调用提示模板
res = tmp.invoke({"name":"Jessica","user_input":"女人比男人长寿吗?"})

model = ChatOpenAI(
        model="qwen2.5vl:latest",                  # 替换为你本地实际的模型名称
        base_url="http://localhost:11434/v1",  # 本地 OpenAI 兼容接口地址
        api_key="ollama",                 # 本地调用无需真实 Key，但参数不可为空
        temperature=0.1,
)

resp = model.invoke(res.to_messages()) # 注意提示模板对象作为模型调用参数的写法
resp.pretty_print()
```

## ChatPromptTemplate的调用方式1，invoke
需要参数： 字典列表，返回值：ChatPromptValue


```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

# 实例化提示模板
tmp = ChatPromptTemplate([
  ("system","你是一个非常友好的Ai助手,你的名字叫{name}"),
  ("user","你好，最近怎么样？"),
  ("ai","我很好，谢谢"),
  ("human","{user_input}")
])

# 调用提示模板
res = tmp.invoke({"name":"Jessica","user_input":"猫的能活多少年?"})
print(res)
model = ChatOpenAI(
        model="qwen2.5vl:latest",                  # 替换为你本地实际的模型名称
        base_url="http://localhost:11434/v1",  # 本地 OpenAI 兼容接口地址
        api_key="ollama",                 # 本地调用无需真实 Key，但参数不可为空
        temperature=0.1,
)

resp = model.invoke(res) # 模板的invoke返回的是一个PromptValue，可以直接传递给模型的invoke方法
resp.pretty_print()
```

## ChatPromptTemplate的调用方式2，format
传入参数：tmp.format(变量名称1=值1，变量名称2=值2，...），返回值：str

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

# 实例化提示模板
tmp = ChatPromptTemplate.from_messages([
  ("system","你是一个非常友好的Ai助手,你的名字叫{name}"),
  ("user","你好，最近怎么样？"),
  ("ai","我很好，谢谢"),
  ("human","{user_input}")
])

# 调用提示模板
prompt_str = tmp.format(name="小明",user_input="牛的能活多少年?") # 返回的是字符串
# print(prompt_str)
model = ChatOpenAI(
        model="gemma4:latest",                  # 替换为你本地实际的模型名称
        base_url="http://localhost:11434/v1",  # 本地 OpenAI 兼容接口地址
        api_key="ollama",                 # 本地调用无需真实 Key，但参数不可为空
        temperature=0.1,
)

# resp = model.invoke([prompt_str]) # 这个prompt_str是一个字符串，必须放在[]里面
resp = model.invoke(prompt_str) # 这个prompt_str是一个字符串，必须放在[]里面
resp.pretty_print()
```

## ChatPromptTemplate调用方法3，
语法： tmp.format_messages(参数1=值1，参数2=值2，...),返回一个信息列表

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

# 实例化提示模板
tmp = ChatPromptTemplate.from_messages([
  ("system","你是一个非常友好的Ai助手,你的名字叫{name}"),
  ("user","你好，最近怎么样？"),
  ("ai","我很好，谢谢"),
  ("human","{user_input}")
])

# 调用提示模板
str_list = tmp.format_messages(name="金元宝",user_input="狗的能活多少年?") # 返回的是字符串
# print(str_list)
model = ChatOpenAI(
        model="gemma4:latest",                  # 替换为你本地实际的模型名称
        base_url="http://localhost:11434/v1",  # 本地 OpenAI 兼容接口地址
        api_key="ollama",                 # 本地调用无需真实 Key，但参数不可为空
        temperature=0.1,
)

resp = model.invoke(str_list) # 这个prompt_str是一个字符串，必须放在[]里面
resp.pretty_print()
```
