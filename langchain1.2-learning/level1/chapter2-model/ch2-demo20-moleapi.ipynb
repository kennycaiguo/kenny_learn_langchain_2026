```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage,SystemMessage
import os

load_dotenv(override=True)
MoleAPI_API_KEY = os.getenv("MoleAPI_API_KEY")
MoleAPI_BASE_URL = os.getenv("MoleAPI_BASE_URL")

llm = ChatOpenAI(
    model="gpt-4o-mini",
    api_key=MoleAPI_API_KEY,
    base_url=MoleAPI_BASE_URL
)

messages = "蛇的英文"

resp = llm.invoke(messages)
resp.pretty_print()
```
